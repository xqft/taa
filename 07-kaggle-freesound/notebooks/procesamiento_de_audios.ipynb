```python
import sys
sys.path.append('../')
from utils.audio import Audio
import os
from pathlib import Path

def process_audios(input_folder, output_folder, target_duration):
    print(f"Comenzando a procesar: {input_folder} hacia {output_folder}")

    audio_files = [f for f in os.listdir(input_folder) if f.endswith('.wav')]
    if not audio_files:
        print(f"No se encontraron archivos .wav en la carpeta: {input_folder}")
        return
    
    print(f"Archivos encontrados: {audio_files}")
    
    for audio_file in audio_files:
        audio_path = Path(input_folder) / audio_file
        if not audio_path.is_file():
            print(f"Archivo no encontrado: {audio_path}")
            continue
        try:
            audio = Audio(str(audio_path), target_duration=target_duration)
            audio.save_spectrogram(output_folder)
        except Exception as e:
            print(f"Error procesando {audio_path}: {e}")

    print(f"Finalizado el procesamiento de: {input_folder}")
```

```python
import os
import librosa
import numpy as np
import matplotlib.pyplot as plt

def process_audio_files(input_dir, output_dir):
    if not os.path.exists(output_dir):
        os.makedirs(output_dir)

    audio_files = [f for f in os.listdir(input_dir) if f.endswith('.wav')]
    print(f"Archivos encontrados: {audio_files}")
    
    for audio_file in audio_files:
        input_path = os.path.join(input_dir, audio_file)
        output_path = os.path.join(output_dir, os.path.splitext(audio_file)[0] + '.png')
        
        if not os.path.exists(input_path):
            print(f"El archivo no existe: {input_path}")
            continue
        
        try:
            print(f"Procesando archivo: {input_path}")
            y, sr = librosa.load(input_path, sr=None)
            
            mel_spec = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=128)
            log_mel_spec = librosa.power_to_db(mel_spec, ref=np.max)
            
            plt.figure(figsize=(10, 4))
            librosa.display.specshow(log_mel_spec, sr=sr, x_axis='time', y_axis='mel')
            plt.axis('off')
            plt.tight_layout(pad=0)
            plt.savefig(output_path, bbox_inches='tight', pad_inches=0)
            plt.close()
            print(f"Espectrograma log-mel guardado en: {output_path}")
            
        except Exception as e:
            print(f"Error procesando el archivo {input_path}: {e}")

input_dir = 'C:/Users/Valen/Desktop/Laboratorio-2-TAA/data/train_curated'
output_dir = 'C:/Users/Valen/Desktop/Laboratorio-2-TAA/data/preproc2'

process_audio_files(input_dir, output_dir)
```

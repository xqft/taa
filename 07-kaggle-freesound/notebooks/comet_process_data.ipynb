```python
import pandas as pd

data = pd.read_json("data_json/train_val_lwlraps.json", orient="records")
data.head()
```

```python
# preprocesar la columna name para poder meterla como label en la grafica.
# esto al final no funciona porque si el nombre es muy largo, Comet lo corta y le pone puntos suspensivos :(
# asi que puse los labels manualmente.

def mod_string(s):
    s = s.lower()
    s = s.replace("lwlrap ", "")
    s = s.replace("entrenamiento", "train")
    s = s.replace("validacion", "val")
    s = s.replace("non-resampled", "")
    return s
    
data["name"] = data["name"].apply(mod_string)
data["name"]
```

```python
import matplotlib.pyplot as plt
import numpy as np

import sys
sys.path.append("../utils")
from plot import config_plot

config_plot()

plt.plot(data.loc[0, "x"], data.loc[0, "y"], label="train")
plt.plot(data.loc[1, "x"], data.loc[1, "y"], label="val")
plt.plot(np.array(data.loc[2, "x"]) + 6, data.loc[2, "y"], label="train fine-tune")
plt.plot(np.array(data.loc[3, "x"]) + 6, data.loc[3, "y"], label="val fine-tune")

plt.xlabel("epoca (desde el primer entrenamiento)")
plt.ylabel("lwlrap")
plt.legend()
plt.show()
```
